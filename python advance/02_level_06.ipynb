# 1. Exceptions

## Topic

Exceptions

## Definition

An **exception** is an event that occurs during program execution and interrupts the normal flow of the program.

Python allows us to handle exceptions instead of allowing the program to terminate unexpectedly.

## Basic Syntax

```python
try:
    risky_code()

except SomeException:
    handle_error()
```

## Example

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)

except ValueError:
    print("Invalid input")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

## How It Works

```text
try
 ↓
Code runs
 ↓
Exception?
 ├── No → continue normally
 └── Yes → matching except block
```

## Important Exception Types

```python
ValueError
TypeError
ZeroDivisionError
IndexError
KeyError
FileNotFoundError
AttributeError
```

## `else`

`else` runs only when the `try` block completes without an exception.

```python
try:
    result = 10 / 2

except ZeroDivisionError:
    print("Error")

else:
    print("Success:", result)
```

## `finally`

`finally` normally executes whether an exception occurs or not.

```python
try:
    file = open("data.txt")

except FileNotFoundError:
    print("File not found")

finally:
    print("Execution finished")
```

## `raise`

`raise` manually creates an exception.

```python
age = 15

if age < 18:
    raise ValueError("Age must be 18 or greater")
```

## Interview Points

**Q: What is exception handling?**
Exception handling is the mechanism used to detect and handle runtime errors using constructs such as `try`, `except`, `else`, and `finally`.

**Q: Difference between `else` and `finally`?**

```text
else
→ runs only when no exception occurs

finally
→ cleanup code that is normally executed regardless of exception
```

**Q: Why use exceptions?**
To handle abnormal situations gracefully and prevent unexpected program termination.

# 6. Generators

## Topic

Generator

## Definition

A **generator** is a special type of iterator created using the `yield` keyword.

Generators produce values lazily, one at a time.

## Example

```python
def numbers():
    yield 1
    yield 2
    yield 3


for number in numbers():
    print(number)
```

Output:

```text
1
2
3
```

## `yield` vs `return`

### `return`

```python
def test():
    return 10
```

Function finishes after returning the value.

### `yield`

```python
def test():
    yield 10
    yield 20
```

Function pauses at each `yield` and continues from there when requested again.

## How It Works

```text
generator()
    ↓
yield 1
    ↓
pause
    ↓
next()
    ↓
yield 2
    ↓
pause
```

## Why Generators?

Generators are useful when processing large amounts of data because they do not need to create the entire sequence in memory at once.

## Example

```python
def numbers():
    for i in range(10_000_000):
        yield i
```

Instead of creating a huge list, values are produced one at a time.

## Important Term

**Lazy Evaluation**

> Values are generated only when they are requested.

# 8. Decorators

## Topic

Decorator

## Definition

A **decorator** is a function that takes another function, wraps it, and adds or modifies its behavior without directly changing the original function.

## Basic Example

```python
def logger(func):

    def wrapper():
        print("Function started")

        func()

        print("Function finished")

    return wrapper


@logger
def greet():
    print("Hello")


greet()
```

Output:

```text
Function started
Hello
Function finished
```

## How It Works

```text
@logger
   ↓
logger(greet)
   ↓
wrapper
   ↓
greet()
```

## Important Concept

A decorator works because:

> Functions are first-class objects in Python.

A function can be:

```text
stored in a variable
passed as an argument
returned from another function
```

## Decorator With Arguments

A practical decorator should usually support `*args` and `**kwargs`.

```python
def logger(func):

    def wrapper(*args, **kwargs):
        print("Function started")

        result = func(*args, **kwargs)

        print("Function finished")

        return result

    return wrapper


@logger
def add(a, b):
    return a + b


print(add(10, 20))
```

Output:

```text
Function started
Function finished
30
```

## Common Real-World Uses

```text
Logging
Authentication
Authorization
Validation
Caching
Performance timing
API routes
Access control
```

## Interview Answer

**Q: What is a decorator?**

> A decorator is a callable that wraps another callable to extend or modify its behavior without modifying its original implementation.

# 9. Closures

## Topic

Closure

## Definition

A **closure** is an inner function that remembers variables from its enclosing scope even after the enclosing function has finished executing.

## Example

```python
def multiplier(x):

    def multiply(y):
        return x * y

    return multiply


double = multiplier(2)

print(double(10))
print(double(20))
```

Output:

```text
20
40
```

## How It Works

```text
multiplier(2)
      ↓
x = 2
      ↓
multiply() created
      ↓
multiply returned
      ↓
double remembers x = 2
```

So:

```python
double(10)
```

acts like:

```text
2 × 10
```

## Another Example

```python
def power(exponent):

    def calculate(number):
        return number ** exponent

    return calculate


square = power(2)
cube = power(3)

print(square(5))
print(cube(5))
```

Output:

```text
25
125
```

Each returned function remembers its own `exponent`.

## Interview Answer

> A closure is an inner function that retains access to variables from its enclosing scope even after the enclosing function has returned.

# 12. Exception Handling Best Practices

### Catch specific exceptions

Good:

```python
try:
    age = int(value)

except ValueError:
    print("Invalid age")
```

Avoid unnecessarily broad handling:

```python
try:
    risky_code()

except Exception:
    pass
```

The latter can hide real bugs.

### Don't silently ignore errors

Bad:

```python
except Exception:
    pass
```

Better:

```python
except Exception as error:
    print(error)
```

In production code, proper logging is generally preferable to plain printing.

# 15. Most Important Interview Questions

### Q1. What is the difference between an error and an exception?

An exception is a runtime event that can be handled by the program. Errors is a broader term and can refer to many kinds of problems, including syntax, runtime, or logical errors.

### Q2. What is the difference between `raise` and `except`?

```text
raise
→ creates/propagates an exception

except
→ catches and handles an exception
```

### Q3. What is a context manager?

> A mechanism for managing resources automatically, especially setup and cleanup, usually through the `with` statement.

### Q4. What is the difference between an iterator and an iterable?

> An **iterable** is an object that can provide an iterator, while an **iterator** is the object that actually produces values using `__next__()`.

Example:

```python
numbers = [1, 2, 3]

iterator = iter(numbers)
```

The list is iterable; `iterator` is the iterator.

### Q5. Why are generators memory efficient?

> They produce values lazily instead of storing the entire sequence in memory at once.

### Q6. What is `yield`?

> `yield` produces a value from a generator and suspends its execution so it can continue later.

### Q7. What is a decorator?

> A decorator wraps a function or other callable to extend or modify its behavior.

### Q8. Why use `*args` and `**kwargs` in decorators?

> To allow the wrapper to accept arbitrary positional and keyword arguments and pass them to the wrapped function.

### Q9. What is a closure?

> A closure is an inner function that retains access to variables from its enclosing scope after the outer function has returned.

### Q10. Are generators iterators?

> Yes. A generator object implements the iterator protocol and can be consumed using `next()`.
