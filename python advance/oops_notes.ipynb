# OOP — Interview Revision Notes

## 1. Class

**Topic:** Class

**Definition:**
A **class** is a blueprint or template used to create objects. It defines the data (attributes) and behavior (methods) that its objects will have.

**Code:**

```python
class Car:
    def start(self):
        print("Car started")
```

**How it works:**
The class itself is only a blueprint. An actual object is created from it.

---

## 2. Object

**Topic:** Object

**Definition:**
An **object** is an instance of a class. It contains actual data and can use the methods defined by the class.

**Code:**

```python
class Car:
    def start(self):
        print("Car started")

car1 = Car()
car1.start()
```

**How it works:**

```text
Car  → Class
car1 → Object
```

One class can create many objects.

---

## 3. Attributes

**Topic:** Attributes

**Definition:**
Attributes are variables that store data associated with an object or class.

**Code:**

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

user = User("Sandeep", 20)

print(user.name)
print(user.age)
```

**How it works:**
`self.name` and `self.age` store data inside the object.

---

## 4. `self`

**Topic:** `self`

**Definition:**
`self` refers to the **current instance/object** of a class.

**Code:**

```python
class User:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello", self.name)

user = User("Sandeep")
user.greet()
```

**How it works:**
When `user.greet()` is called, `self` refers to `user`.

---

## 5. Constructor / `__init__`

**Topic:** Constructor (`__init__`)

**Definition:**
`__init__()` is the initializer method that runs automatically when an object is initialized. It is commonly called the constructor in beginner Python terminology.

**Code:**

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

user = User("Sandeep", 20)
```

**How it works:**

```text
User("Sandeep", 20)
        ↓
   __init__()
        ↓
name = "Sandeep"
age = 20
```

---

# 6. Instance Method

**Topic:** Instance Method

**Definition:**
An instance method is a method that works with the data of a specific object. It receives `self` as its first parameter.

**Code:**

```python
class User:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello", self.name)

user = User("Sandeep")
user.greet()
```

**How it works:**
`self` gives the method access to the current object's attributes.

---

# 7. Class Method

**Topic:** Class Method

**Definition:**
A class method works with the class itself rather than a particular object. It uses `@classmethod` and receives `cls`.

**Code:**

```python
class User:
    total_users = 0

    def __init__(self, name):
        self.name = name
        User.total_users += 1

    @classmethod
    def get_total_users(cls):
        return cls.total_users

User("Sandeep")
User("Rahul")

print(User.get_total_users())
```

**How it works:**
`cls` refers to the class.

```text
self → object
cls  → class
```

---

# 8. Static Method

**Topic:** Static Method

**Definition:**
A static method is a method that does not need access to instance or class data. It uses `@staticmethod`.

**Code:**

```python
class MathUtils:

    @staticmethod
    def add(a, b):
        return a + b

print(MathUtils.add(10, 20))
```

**How it works:**
It behaves like a normal function placed inside a class for logical organization.

---

# 9. Instance vs Class vs Static Method

**Topic:** Method Types

**Definition:**
Python classes can have instance methods, class methods, and static methods, each serving a different purpose.

**Code:**

```python
class Example:

    def instance_method(self):
        print("Instance")

    @classmethod
    def class_method(cls):
        print("Class")

    @staticmethod
    def static_method():
        print("Static")
```

**How it works:**

```text
Instance Method
→ self
→ works with object data

Class Method
→ cls
→ works with class data

Static Method
→ no self / cls
→ independent utility logic
```

---

# 10. Inheritance

**Topic:** Inheritance

**Definition:**
Inheritance allows a child class to reuse attributes and methods from a parent class.

**Code:**

```python
class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Barking")


dog = Dog()

dog.eat()
dog.bark()
```

**How it works:**

```text
Animal
   ↓
  Dog
```

`Dog` inherits `eat()` from `Animal`.

---

# 11. Method Overriding

**Topic:** Method Overriding

**Definition:**
Method overriding occurs when a child class provides its own implementation of a method inherited from the parent class.

**Code:**

```python
class Animal:

    def sound(self):
        print("Some sound")


class Dog(Animal):

    def sound(self):
        print("Bark")


dog = Dog()
dog.sound()
```

**How it works:**
Python uses the child class's version of `sound()` for a `Dog` object.

---

# 12. `super()`

**Topic:** `super()`

**Definition:**
`super()` is used to access methods or the initializer of a parent class from a child class.

**Code:**

```python
class Animal:

    def __init__(self, name):
        self.name = name


class Dog(Animal):

    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed


dog = Dog("Tommy", "Labrador")
```

**How it works:**

```text
Dog.__init__()
      ↓
super().__init__()
      ↓
Animal.__init__()
```

---

# 13. Encapsulation

**Topic:** Encapsulation

**Definition:**
Encapsulation means bundling data and methods together inside a class and controlling how internal state is accessed or modified.

**Code:**

```python
class BankAccount:

    def __init__(self, balance):
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount

    def get_balance(self):
        return self._balance


account = BankAccount(1000)
account.deposit(500)

print(account.get_balance())
```

**How it works:**
The object controls access to its internal state through methods.

### Important

```python
self._balance
```

means internal-use-by-convention.

```python
self.__balance
```

uses Python's name-mangling mechanism.

---

# 14. Abstraction

**Topic:** Abstraction

**Definition:**
Abstraction hides implementation details and exposes only the essential interface needed by the user.

**Code:**

```python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Bark")
```

**How it works:**
The abstract class says:

```text
Every Animal must provide sound()
```

The child class decides **how** `sound()` works.

---

# 15. Polymorphism

**Topic:** Polymorphism

**Definition:**
Polymorphism means the same interface or method name can produce different behavior depending on the object.

**Code:**

```python
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")


animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()
```

**How it works:**

```text
Dog.sound() → Bark
Cat.sound() → Meow
```

The caller only needs the common interface `sound()`.

---

# 16. Composition

**Topic:** Composition

**Definition:**
Composition means building a class by containing objects of other classes. It represents a **HAS-A** relationship.

**Code:**

```python
class Engine:

    def start(self):
        print("Engine started")


class Car:

    def __init__(self):
        self.engine = Engine()

    def start(self):
        self.engine.start()


car = Car()
car.start()
```

**How it works:**

```text
Car
 ↓
HAS-A
 ↓
Engine
```

### Remember

```text
IS-A  → Inheritance
HAS-A → Composition
```

---

# 17. Magic / Dunder Methods

**Topic:** Magic Methods

**Definition:**
Magic methods, also called dunder methods, are special methods whose names begin and end with double underscores. They define how objects behave with Python's built-in operations.

**Code:**

```python
class User:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


user = User("Sandeep")

print(user)
```

**How it works:**
When `print(user)` is executed, Python uses `__str__()` to get the object's string representation.

### Common examples

```python
__init__   → initialize object
__str__    → string representation
__len__    → len(object)
__add__    → object + object
__eq__     → object == object
```

# 18. Dataclasses

**Topic:** Dataclasses

**Definition:**
A dataclass is a class designed primarily for storing data. The `dataclasses` module can automatically generate common methods and reduce boilerplate code.

**Code:**

```python
from dataclasses import dataclass


@dataclass
class User:
    name: str
    age: int


user = User("Sandeep", 20)

print(user)
```

Output:

```text
User(name='Sandeep', age=20)
```

**How it works:**
Python generates useful methods such as `__init__()` and `__repr__()` automatically.

# 19. The Four Pillars of OOP

## 19.1 Encapsulation

**Topic:** Encapsulation

**Definition:**
Keep data and related methods together and control access to internal state.

**Code:**

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance
```

**Interview keyword:**
**Data hiding / controlled access**

---

## 19.2 Abstraction

**Topic:** Abstraction

**Definition:**
Expose essential behavior while hiding unnecessary implementation details.

**Code:**

```python
from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class UPI(Payment):

    def pay(self, amount):
        print(f"Paid {amount}")
```

**Interview keyword:**
**Hide implementation details, expose interface**

---

## 19.3 Inheritance

**Topic:** Inheritance

**Definition:**
A child class acquires or reuses behavior from a parent class.

**Code:**

```python
class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):
    pass


dog = Dog()
dog.eat()
```

**Interview keyword:**
**Code reuse / IS-A relationship**

---

## 19.4 Polymorphism

**Topic:** Polymorphism

**Definition:**
The same method/interface can behave differently for different objects.

**Code:**

```python
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")


for animal in [Dog(), Cat()]:
    animal.sound()
```

**Interview keyword:**
**One interface, many implementations**

---

# 20. Class Variable vs Instance Variable

**Topic:** Class Variable vs Instance Variable

**Definition:**
An instance variable belongs to a particular object, while a class variable is shared at the class level.

**Code:**

```python
class Student:

    school = "ABC School"          # class variable

    def __init__(self, name):
        self.name = name            # instance variable


s1 = Student("Sandeep")
s2 = Student("Rahul")

print(s1.name)
print(s2.name)

print(s1.school)
print(s2.school)
```

**How it works:**

```text
name   → different for each object
school → shared class-level value
```
