```python
# comprehensions 
number = [1,2,3,4,5,]

squr = [num ** 2 for num in number]
print(squr)


# with conditions 
numbers = [1,2,3,4,5,6]

result = [num ** 2 for num in numbers if num % 2 == 0]
print(result)


#with strings
names = ["sandeep", "rahul", "amit"]
upper_names = [name.upper() for name in names]
print(upper_names)

# with list 
ages = [12, 18, 25, 15, 30, 16]
adults = [age for age in ages if age >= 18]
print(adults)


# with dict 
numbers = [1, 2, 3, 4, 5]
squares = {num: num ** 2 for num in numbers}
print(squares)

# with sets 
numbers = [1, 2, 2, 3, 3, 4]
unique = {num for num in numbers}
print(unique)

# nested loop 
result = [(i, j) for i in range(3) for j in range(3)]
print(result)
```

```python
# unpacking 
numbers = [1, 2, 3, 4, 5]

a, *b = numbers

print(a)
print(b)

# direct_unpacking 
a, b = [10, 20]

print(a)  # 10
print(b)  # 20


# tuple_unpacking 
person = ("Sandeep", 20, "Python")

name, age, language = person

print(name)
print(age)
print(language)
```

```python
# dict_unpacking
person = {
    "name": "Sandeep",
    "age": 20
}

print(person)

def introduce(name, age):
    print(name, age)

introduce(**person)
```

```python
# *args 
def add(*numbers):
    print(numbers)

add(10, 20, 30, 40) #return tuple

def add_num(*numbers):
    total = 0
    for num in numbers:
        total += num

    return total

add_num(10, 20, 30, 40)
```

```python
# decorator 
import time
from functools import wraps

def measure_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"⏱️ '{func.__name__}' took {end_time - start_time:.4f} seconds to run.")
        return result
    return wrapper

@measure_time
def heavy_calculation():
    time.sleep(1.5)  # Simulates a slow process

heavy_calculation()

```
