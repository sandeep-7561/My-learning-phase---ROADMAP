```python
# method of list 
# Initialize a list
fruits = ["apple", "banana", "cherry"]

# 1. Add elements
fruits.append("orange")        # ['apple', 'banana', 'cherry', 'orange']
fruits.insert(1, "mango")      # ['apple', 'mango', 'banana', 'cherry', 'orange']
fruits.extend(["grape", "kiwi"]) # Adds multiple items to the end

# 2. Search and count
print(fruits.index("banana"))  # Output: 2
print(fruits.count("apple"))   # Output: 1

# 3. Remove elements
fruits.remove("mango")         # Removes "mango"
popped_item = fruits.pop()     # Removes and returns "kiwi"

# 4. Reorder
fruits.sort()                  # Sorts alphabetically in place
fruits.reverse()               # Reverses the order in place

# 5. Duplicate and empty
fruits_copy = fruits.copy()    # Creates a separate copy
fruits.clear()                 # Leaves fruits as []
```

```python
# comprehensions 
number = [1,2,3,4,5,]

squr = [num ** 2 for num in number]
print(squr)


# with conditions 
numbers = [1,2,3,4,5,6]

result = [num ** 2 for num in numbers if num % 2 == 0]
print(result)


#with strings
names = ["sandeep", "rahul", "amit"]
upper_names = [name.upper() for name in names]
print(upper_names)

# with list 
ages = [12, 18, 25, 15, 30, 16]
adults = [age for age in ages if age >= 18]
print(adults)


# with dict 
numbers = [1, 2, 3, 4, 5]
squares = {num: num ** 2 for num in numbers}
print(squares)

# with sets 
numbers = [1, 2, 2, 3, 3, 4]
unique = {num for num in numbers}
print(unique)

# nested loop 
result = [(i, j) for i in range(3) for j in range(3)]
print(result)
```
