```python
from typing import Optional, Union, Literal
def greet(name: str) -> Optional[str]:
    return f"Hello {name}"

print(greet('Sandeep'))

def payment_process(payment: Union[int,float]) -> None:
    return payment

print(payment_process(10))

from typing import Literal

def set_theme(mode: Literal["light", "dark", "system"]) -> None:
    # The IDE will warn you if you pass "blue" or "DARK" (case-sensitive)
    print(f"Setting theme to {mode}")

set_theme('light')
```

```python
# Typehints 
name: str = 'Sandeep'
age: int = 20
payment: float = 5000.2
is_active: bool = True

print(name,age,payment,is_active)

def add(x: int, y: int) -> int:
    return x + y

print(add(5,5))
```

```python
# Optional 
'''
Optional[T] means a value can either be of type T or None.
'''
from typing import Optional

def find_user(user_id: int) -> Optional[str]:
    if user_id == 1:
        return "Sandeep"

    return None

print(find_user(1))

def find_user_1(usr_id: int) -> str | None:
    return f"Hello user, Your id is {usr_id}"

print(find_user_1(55))
```

```python
# union 
from typing import Union

def process(value: Union[int,str]) -> str:
    return str(value)

print(process(5))
print(process('Hello'))


def process_new(value: int | str) -> str:
    return str(value)
print(process_new(555))
```

```python
# Literal 
from typing import Literal

def run(option: Literal['fast','slow']) -> str:
    return f"Your run {option}"

print(run('fast'))
```

```python
# Generics 
from typing import TypeVar

T = TypeVar('T')

def first(items: list[T]) -> T:
    return items[0]

list1 = [1,2,3,4,5]
list2 = ['sandeep','simran']
print(first(list1))
print(first(list2))
```

```python
# Generic Classes 
''' 
A generic class can work with different data types while keeping type information consistent.
'''

from typing import Generic, TypeVar

T = TypeVar("T")


class Box(Generic[T]):

    def __init__(self, value: T):
        self.value = value

    def get(self) -> T:
        return self.value

number_box = Box[int](100)
text_box = Box[str]("Hello")
```

```python
# protocol 
from typing import Protocol

# 1. Define the structural contract
class Flyer(Protocol):
    def fly(self) -> str:
        ...  # Use ellipsis as a placeholder for methods

# 2. These classes DO NOT explicitly inherit from Flyer
class Airplane:
    def fly(self) -> str:
        return "Roaring through the clouds"

class Bird:
    def fly(self) -> str:
        return "Flapping wings in the sky"

class Car:
    def drive(self) -> str:
        return "Driving on the road"

# 3. Use the Protocol as a type hint
def launch(vehicle: Flyer):
    print(vehicle.fly())

# These are perfectly valid according to static type checkers
launch(Airplane()) 
launch(Bird())     

```

```python
# Callable 
from collections.abc import Callable

def process(func: Callable[[int], int], value: int) -> int:
    return func(value)

def square(x: int) -> int:
    return x * x

print(process(square, 5))
```

```python
# Final 
from typing import Final

max_ret = Final = 3
print(max_ret)
print(type(max_ret))
```

```python
# Type Narrowing 
def process(value: int | str):
    if isinstance(value,int):
        print(value+10)
    else:
        print(value.upper())

process('sjdksj')
```
